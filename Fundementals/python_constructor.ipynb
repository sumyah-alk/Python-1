In Python, there are two kinds of attributes:
1. Instance Attributes

Belong to a specific object (instance), not the class itself.

Usually defined inside __init__ using self.

Example:

class Dog:
    def __init__(self, name, age):
        self.name = name  # instance attribute
        self.age = age    # instance attribute

dog1 = Dog("Buddy", 5)
dog2 = Dog("Max", 3)


dog1.name → "Buddy"

dog2.name → "Max"

Even though the class is Dog, self.name and self.age belong to each instance.
Changing dog1.name does not change dog2.name.

2. Class Attributes
Belong to the class itself, shared across all instances.

Defined directly in the class body, not inside __init__.

Example:

class Dog:
    species = "Canis familiaris"  # class attribute

dog1 = Dog("Buddy", 5)
dog2 = Dog("Max", 3)

print(dog1.species)  # Canis familiaris
print(dog2.species)  # Canis familiaris


species is shared. If you change Dog.species, it affects all instances that haven’t overridden it.
