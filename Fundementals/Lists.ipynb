#### Exercise: Average Calculator with Sentinel Value

Write a Python program that repeatedly asks the user to enter numbers, one at a time. The program should keep track of a running total and a count of how many numbers were entered.

The user can enter as many numbers as they want.
When the user is done entering numbers, they should type the word "done" instead of a number. This should stop the input loop.
Once the loop stops, calculate and print the average of all the numbers the user entered.

```python
total=0
count=0
while True:
    inp = input('Enter a number: ')
    if inp == 'done' : break # single line if used for one action 
    value = float(inp)
    total = total + value
    count = count + 1
   
avg=total/count
print(total)
print(avg)
# this code says loop infinitely and unless something inside it explicitly stops it (via break).
#This is a common pattern when you don't know in advance how many times you'll loop
#it will check if the val is done then directly leave the loop 
#if if is not done then it will go the next line 
```

```python
total=0
count=0
li=list()
while True:
    inp = input('Enter a number: ')
    if inp == 'done' : break # single line if used for one action 
    value = float(inp)
    li.append(value)

avg=sum(li)/len(li)
print(avg)
```

## Strip() vs Split() 
#### strip() cleans the ends of a string and returns a string. split() breaks a string into pieces and returns a list.

```python
c = "  hello my friend  \n"
print(c.strip())
#It removes whitespace (spaces, tabs, \n) from both ends only.
#  Spaces in the middle are untouched, and the result is still one string.

c=c.split()
print(c)
# ['hello', 'my', 'friend']
```
