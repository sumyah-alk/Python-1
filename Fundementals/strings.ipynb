Looping through strings

```python
name="hello"
i=0
while i<len(name):#traverse the index until we reach the value length of the string then leave
	print(name[i], end="") 
	i+=1
print()

#another way
i=0
while i < len(name):
	letter=name[i]
	print(i , letter)
	i=i+1
```

A definite loop (in Python, the for loop) is used when you know in advance how many times you want to repeat something — i.e., the number of iterations is fixed or determined by something you can count, like the length of a list or a specific range of numbers.

Rule of thumb:

Know the count/collection ahead of time → definite loop (for)
Depends on a condition that changes during execution → indefinite loop (while)

Advise : always use the simplest way to do something 
so in case you have a collection and you want to just print it then the definite loop is better for readability

```python
letter="hello"
for i in letter:
	print(i,end="")
```

## Python String find() Method 
syntax : string.find(value, start, end)


```python
txt = "Hello, welcome to my world."
print("the address before re-assign ", id(txt))
print(txt)

txt = txt.find("Hello") #it returns the first occurence index

print(txt)
print("the address after re-assign ", id(txt))

#What's happening step by step:

# txt = "Hello, welcome to my world." → txt points to a string object at some memory address (let's call it Address A).
# txt = txt.find("Hello") → .find() returns an integer (the index, 0 in this case). Now txt is reassigned to point to a new object — an integer — at a different address (Address B).
# The original string object at Address A is no longer referenced by anything (assuming nothing else points to it).
# Python's garbage collector (specifically, reference counting, which is Python's main GC mechanism) notices the reference count for that string object dropped to 0, and it gets deallocated — the memory is freed.


print("*************")
txtt = "Hello, welcome to my world."


x = txtt.find("z") #if not found it returns -1
print(x)
#If the value is not found, the find() method returns -1, but the index() method will raise an exception:

x=txtt.find("l",6,16)
print(x)


```

### strip() Method
strip() method removes specified characters from both the beginning and the end of a string. If no characters are provided, it removes leading and trailing whitespace by default.

```python
prog=" python "
x=prog.strip()
print(x)
```

string.strip(chars)


Parameters: chars (optional) - Specifies the characters to remove from the beginning and end of the string. If omitted, whitespace characters are removed.
Return Type: A new string with the specified characters removed from both ends. The original string remains unchanged.

```python
text="((((hello))))"
res=text.strip("()")
print(res)#Explanation: strip("()") removes all ( and ) characters from both ends. Characters inside the string are not affected.

gg=" lll"
x=gg.lstrip() #removes whitespace on the left
print(x)
ggg="lll  "
x=gg.rstrip() #removes whitespace on the lefright
print(x)


```

```python
date = "23-09-2026"
date=date[6:]
print(date)
print("**************")
#but what it is a very long string , it is not practical to keep counting the start index to start slicing
ii=date.find("2026") #2 start
#how to get the end index
goali=ii+4 # till will read until the index of 6 
while ii != goali :
	print(date[ii],end="")
	ii+=1
     
print("**************")
#more enhanced 
date = "23-09-2026"

ii = date.find("2026")      # search the ORIGINAL string, not the sliced one
goali = ii + len("2026")     # dynamic length instead of hardcoded +4

while ii < goali:
    print(date[ii], end="")
    ii += 1
```



why the *** are beside the 2026 although i did the print(***) outside the while loop ?
print()'s newline (the thing that moves output to a new line) is controlled by the end parameter, not by your code's structure.

Inside the loop, you wrote print(date[ii], end=""), which tells Python "don't add a newline after this." So after all 4 loop iterations, the output cursor is still sitting right after 2026, on the same line.

Being "outside the while loop" only changes which code runs next — it doesn't insert a newline for you. So when print("**************") runs next, it just continues writing from wherever the cursor already was (right after 2026), giving you 2026************** glued together on one line.
