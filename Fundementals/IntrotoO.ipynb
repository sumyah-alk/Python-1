##Way 1
def print_items(n): - Here, we're defining a function named print_items that takes one parameter, n.

for i in range(n): - This is a for loop. It will iterate over a sequence of numbers generated by the function range(n). The range(n) function generates a sequence of numbers from 0 up to but not including n. For each iteration, i takes the next value in this sequence.

print(i) - This is a function call to print, which outputs the current value of i to the console.

print_items(10) - This is a call to the function print_items with the argument 10. This will result in the function running its code (the loop) with n being 10, and so it will print numbers from 0 up to but not including 10.

as you can see the O(n) is directly propotional with the input of data size n with number of operations


```python
#using while loop 

def print_items(n):
	i=0
	while i<n:
		print(i)
		i+=1
print_items(10)
```

```python
#Using a for loop with an explicitly defined list:
def print_items(n):
	for i in list(range(n)):
		print(i)

print_items(10)
```

Default constructor

If you don’t define an __init__ method, Python automatically gives the class a default constructor.

The default constructor takes no arguments (except self, which Python handles automatically).


Example:

```python
class Cookie: 
	def __init__(self,color):
		self.color=color
	def get_colors(self):
		return self.color
	
cookie1=Cookie('green')
p=cookie1.get_colors()
print(p)

```

# Creating constructor for linkedlist


# LinkedList

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None


class LinkedList:  # we capitalize the name of the class
    def __init__(self, value):  # the constructor will create a new node
        new_node = Node(value)
        self.head = new_node
        self.tail = new_node
        self.length = 1


my_linked_list = LinkedList(4)

print(my_linked_list.head.value)  # Output: 4
print(my_linked_list.length)      # Output: 1
```
